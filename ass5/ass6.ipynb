```python
import cv2
import numpy as np
```

```python
image = cv2.imread("img.jpg",cv2.IMREAD_GRAYSCALE)
```

```python
cv2.imshow('My Window', image)   # window_name, image
cv2.waitKey(0)                  # waits until a key is pressed
cv2.destroyAllWindows()   
```

```python
arr = np.zeros(256)

for i in range(0,256):
    arr[i] = np.count_nonzero(image == i)  

(rows,column)= image.shape
totalSize = rows*column
```

```python
P_r = np.zeros(256)
for i in range(0 , 256):
    P_r[i] = (arr[i]/totalSize)

Cumulative = np.zeros(256)
summ = 0
for i in range(0,256):
    summ += P_r[i]
    Cumulative[i] = summ
print(Cumulative)
```

```python
for i in range(0 , rows):
    for j in range(0,column):
        ele = image[i][j]
        newVal = 255*(Cumulative[ele])
        image[i][j] = newVal
    
```

```python
cv2.imshow("img" , image)
cv2.waitKey(0)
```
