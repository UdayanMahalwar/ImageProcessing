```python
import cv2
import numpy as np
```

```python
img = cv2.imread("img6.jpg")
print(img.shape)
rows,column,length= img.shape
for i in range(1,rows-1):
    for j in range(1,column-1):
        ridx = i
        cidx = j
        for k in range(3):
            img[i][j][k] = (np.sum(np.array([img[ridx-1 ,cidx , k] , img[ridx+1 , cidx , k] , img[ridx , cidx+1 , k] , img[ridx , cidx-1 , k],img[ridx-1,cidx-1,k] , img[ridx-1, cidx+1 , k] , img[ridx+1 , cidx-1 , k] , img[ridx+1 , cidx+1 , k]]))/6);
cv2.imshow("img6.jpg" , img)
cv2.waitKey(0)
```
