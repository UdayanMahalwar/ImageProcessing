```python
import cv2 as cv2
import numpy as np
pixel = np.array(cv2.imread("image.jpg" , 0));
pixel = pixel.astype('uint8')
sz = pixel.shape;

print(pixel[0]);
c = 255/np.log(1+255);
for i in range(0,sz[0]) :
    for j in range(0 , sz[1]):
            val = pixel[i][j];
            s = c*(np.log(1 + val));
            pixel[i,j] = np.uint8(s)
print(pixel);
cv2.imshow("converted" ,pixel)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
